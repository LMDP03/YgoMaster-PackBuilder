```python
import requests
import csv

PACK_ID = 24110

url = f"https://ygoprodeck.com/api/pack-sim/getPackInfo.php?id={PACK_ID}"

response = requests.get(url, timeout=30)
response.raise_for_status()
data = response.json()

print("Chiavi disponibili:", list(data) if isinstance(data, dict) else "Lista")

# Individua la lista delle carte
if isinstance(data, list):
    cards = data
else:
    cards = next(
        (
            value for key, value in data.items()
            if key.lower() in ("cards", "cardlist", "card_list", "packcards")
            and isinstance(value, list)
        ),
        None
    )

if cards is None:
    raise ValueError("Lista carte non trovata: controllare la struttura JSON")

with open(f"pack_{PACK_ID}.csv", "w", newline="", encoding="utf-8-sig") as f:
    writer = csv.writer(f)
    writer.writerow(["Card ID", "Name", "Rarity"])

    for card in cards:
        writer.writerow([
            card.get("name", ""),
            card.get("card_rarity", "")
        ])

print("CSV creato!")
```
